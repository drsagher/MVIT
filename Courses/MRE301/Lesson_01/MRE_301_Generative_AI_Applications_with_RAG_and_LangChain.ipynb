```python
from pptx import Presentation
from langchain_core.documents import Document

prs = Presentation("sample_data/AI _Agents_ RAG.pptx")
documents = []

for slide_number, slide in enumerate(prs.slides, start=1):
    texts = []

    for shape in slide.shapes:
        # Text boxes, titles, placeholders
        if shape.has_text_frame:
            for paragraph in shape.text_frame.paragraphs:
                line = "".join(run.text for run in paragraph.runs).strip()
                if line:
                    texts.append(line)

        # Tables
        if shape.has_table:
            for row in shape.table.rows:
                texts.append(" | ".join(cell.text for cell in row.cells))

    # Speaker notes
    notes = ""
    if slide.has_notes_slide:
        notes = slide.notes_slide.notes_text_frame.text

    documents.append(
        Document(
            page_content="\n".join(texts),
            metadata={
                "source": "marketing-deck.pptx",
                "slide": slide_number,
                "notes": notes,
            },
        )
    )

print(documents[0])
```

**Build a simple multi-format loader**

```python
from pathlib import Path

LOADERS = {
    ".txt":  TextLoader,
    ".pdf":  PyMuPDFLoader,
    ".md":   UnstructuredMarkdownLoader,
    ".csv":  CSVLoader,
    ".docx": Docx2txtLoader,
    ".xlsx": UnstructuredExcelLoader,
    ".pptx": UnstructuredPowerPointLoader,
}

def load_any(path):
    ext = Path(path).suffix.lower()
    loader_cls = LOADERS.get(ext, UnstructuredFileLoader)   # fallback
    return loader_cls(str(path)).load()

# Ensure the directories exist
client_files_dir = Path("client_files")
client_files_dir.mkdir(exist_ok=True)
sample_data_dir = Path("sample_data")

docs = []

# Load files from client_files directory
for file in client_files_dir.iterdir():
    if file.is_file():
        try:
            docs.extend(load_any(file))
        except Exception as e:
            print(f"Could not load {file.name}: {e}")

# Load files from sample_data directory
if sample_data_dir.exists():
    for file in sample_data_dir.iterdir():
        if file.is_file():
            try:
                docs.extend(load_any(file))
            except Exception as e:
                print(f"Could not load {file.name}: {e}")

print(f"Loaded {len(docs)} documents")
```
